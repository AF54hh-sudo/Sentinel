# 01 — AMX Tech Data Validation
Reproducible Phase 4 validation and cleaning review. Ground-truth scenarios are never loaded.

```python
from pathlib import Path

from sentinel.data.loader import read_csv_dataset
from sentinel.data.validation import validate_dataset
from sentinel.eda.cleaning import clean_dataset
from sentinel.eda.summary import profile_dataset

tables = read_csv_dataset(Path('../data/generated'))
raw_report = validate_dataset(tables)
raw_report.row_counts, raw_report.warnings
```

```python
cleaned, cleaning_report = clean_dataset(tables)
{
    'rows_before': cleaning_report.rows_before,
    'rows_after': cleaning_report.rows_after,
    'duplicate_events_removed': cleaning_report.duplicate_events_removed,
    'industries_filled': cleaning_report.industries_filled,
    'satisfaction_retained_missing': cleaning_report.satisfaction_values_retained_missing,
}
```

```python
profiles = profile_dataset(cleaned)
profiles['support_tickets']['dataset'], profiles['support_tickets']['missing']
```

## Cleaning rationale
Unknown industries are explicitly labeled. Duplicate ticket events are removed using all business fields except the surrogate ticket ID. Missing satisfaction is retained because non-response is not neutral satisfaction. Active-subscription end dates and open-ticket resolution dates remain structurally null.
