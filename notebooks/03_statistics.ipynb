# 03 — AMX Tech Statistical Inference
Phase 6 analysis with explicit hypotheses, test selection, effect sizes, assumptions, and causal limitations.

```python
from sentinel.config import get_settings
from sentinel.database.connection import create_database_engine
from sentinel.database.loader import read_database_tables
from sentinel.eda.cleaning import clean_dataset
from sentinel.statistics.business_analysis import run_business_statistics

engine = create_database_engine(get_settings().database_url)
try:
    tables, cleaning_report = clean_dataset(read_database_tables(engine))
finally:
    engine.dispose()
report = run_business_statistics(tables)
report['assumption_diagnostics']
```

```python
report['descriptive_statistics'], report['confidence_intervals']
```

```python
for analysis_name, result in report['tests'].items():
    print('\n', analysis_name)
    print('Test:', result['test_used'])
    print('Reason:', result['reason_for_test'])
    print('n:', result['sample_size'])
    print('Statistic:', result['statistic'], 'p:', result['p_value'])
    print(result['effect_size_name'], result['effect_size'])
    print(result['business_interpretation'])
    print('Limitations:', result['limitations'])
```

## Interpretation boundary
Support analyses use one row per customer and censor tickets at first cancellation. Welch estimates a mean difference; Mann–Whitney provides a robust distributional comparison. Spearman is primary for skewed resolution and bounded satisfaction, while Pearson is a secondary linear sensitivity analysis. None of these tests establishes causation.
